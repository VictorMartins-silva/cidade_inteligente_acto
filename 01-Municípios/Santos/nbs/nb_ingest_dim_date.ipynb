```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from datetime import datetime, timedelta

start_date = "2020-01-01"
end_date = "2030-12-31"

date_range = spark.sql(f"""
    SELECT explode(sequence(to_date('{start_date}'), to_date('{end_date}'), interval 1 day)) as date
""")

meses_pt = ["janeiro", "fevereiro", "março", "abril", "maio", "junho", 
            "julho", "agosto", "setembro", "outubro", "novembro", "dezembro"]

dias_pt = ["domingo", "segunda-feira", "terça-feira", 
           "quarta-feira", "quinta-feira", "sexta-feira", "sábado"]

# monta a string "'janeiro','fevereiro',...'dezembro'"
meses_expr = ",".join([f"'{m}'" for m in meses_pt])
dias_expr = ",".join([f"'{d}'" for d in dias_pt])

date_dimension = date_range.select(
    col("date").alias("full_date"),
    year("date").alias("year"),
    month("date").alias("month_number"),
    expr(f"element_at(array({meses_expr}), month(date))").alias("month_name"),
    dayofmonth("date").alias("day_of_month"),
    dayofweek("date").alias("day_of_week"),
    expr(f"element_at(array({dias_expr}), dayofweek(date))").alias("day_name")
)

date_dimension.write.mode("overwrite").option("overwriteSchema", "true").format("delta").saveAsTable("dim_date_1")
date_dimension.write.mode("overwrite").option("overwriteSchema", "true").format("delta").saveAsTable("dim_date_2")
```
