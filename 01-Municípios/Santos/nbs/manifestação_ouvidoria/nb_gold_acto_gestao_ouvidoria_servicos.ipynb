```python
from pyspark.sql import DataFrame
from functools import reduce

sepref = spark.table("gold_sepref_servicos")
seinfra = spark.table("gold_seinfra_servicos")
cet = spark.table("gold_cet_servicos")
segov = spark.table("gold_segov_servicos")
manif = spark.table("gold_manifestacoes_ouvidoria")

df_unificado = reduce(
    DataFrame.unionAll, 
    [sepref, seinfra, cet, segov, manif]
)

(
    df_unificado
    .write.mode("overwrite")
    .format("delta")
    .option("overwriteSchema", "true")
    .saveAsTable("gold_ouvidoria_servicos")
)
```
