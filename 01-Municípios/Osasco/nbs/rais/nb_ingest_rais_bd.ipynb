```python
%pip install google-cloud-bigquery pyarrow db-dtypes

from google.cloud import bigquery
import pandas as pd

# Credenciais Big Query
credentials_path = "/lakehouse/default/Files/bd2024-444413-1084f2b9d765.json"

client = bigquery.Client.from_service_account_json(credentials_path)

# Query
query = """
SELECT
    dados.ano as ano,
    dados.sigla_uf as sigla_uf,
    dados.id_municipio as id_municipio,
    dados.quantidade_vinculos_ativos as quantidade_vinculos_ativos,
    dados.quantidade_vinculos_clt as quantidade_vinculos_clt,
    dados.quantidade_vinculos_estatutarios as quantidade_vinculos_estatutarios,
    dados.tamanho_estabelecimento as tamanho_estabelecimento,
    dados.cnae_1 as cnae_1,
    dados.cnae_2 as cnae_2,
    dados.cnae_2_subclasse as cnae_2_subclasse
FROM `basedosdados.br_me_rais.microdados_estabelecimentos` AS dados
WHERE dados.sigla_uf = 'SP'
AND dados.ano >= 1993
"""

df = client.query(query).to_dataframe()

df_grp = df.groupby([
    "ano", "sigla_uf", "id_municipio",
    "tamanho_estabelecimento", "cnae_1", "cnae_2",
    "cnae_2_subclasse"
], as_index=False).agg({
    "quantidade_vinculos_ativos": "sum",
    "quantidade_vinculos_clt": "sum",
    "quantidade_vinculos_estatutarios": "sum"
})

# Conversão spark + save na bronze
spark_df = spark.createDataFrame(df_grp)
spark_df.write.mode("overwrite").format("delta").save("Tables/raw_rais_estab_sp")
spark.sql("""
CREATE OR REPLACE TABLE raw_rais_estab_sp
USING DELTA
AS SELECT * FROM delta.`Tables/raw_rais_estab_sp`
""")
```
