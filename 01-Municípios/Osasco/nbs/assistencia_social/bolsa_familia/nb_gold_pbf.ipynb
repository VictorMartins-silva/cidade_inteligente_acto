```python
%%sql
CREATE OR REPLACE TABLE gold_pbf_municipios_selecionados AS
SELECT 
    -- CORREÇÃO AQUI: Mudamos o padrão para ler '2023-12-01' corretamente
    TO_DATE(CAST(mes_referencia AS STRING), 'yyyy-MM-dd') AS mes_referencia,

    CASE 
        WHEN nome_municipio = 'OSASCO' THEN 'Osasco (SP)'
        WHEN nome_municipio = 'SAO BERNARDO DO CAMPO' THEN 'São Bernardo do Campo (SP)'
        WHEN nome_municipio = 'SOROCABA' THEN 'Sorocaba (SP)'
        WHEN nome_municipio = 'RIBEIRAO PRETO' THEN 'Ribeirão Preto (SP)'
        WHEN nome_municipio = 'SANTO ANDRE' THEN 'Santo André (SP)'
        WHEN nome_municipio = 'SAO JOSE DOS CAMPOS' THEN 'São José dos Campos (SP)'
        ELSE INITCAP(nome_municipio) 
    END AS nome_municipio,

    SUM(valor_parcela) AS total_repasses,
    COUNT(DISTINCT nis_favorecido) AS n_favorecidos,
    ROUND(AVG(valor_parcela), 2) AS media_repasses

FROM lh_cidade_inteligente_osasco.silver_pbf_sp

WHERE nome_municipio IN (
    'OSASCO', 
    'SAO BERNARDO DO CAMPO', 
    'SOROCABA', 
    'RIBEIRAO PRETO', 
    'SANTO ANDRE', 
    'SAO JOSE DOS CAMPOS'
)

GROUP BY 
    1, 2
ORDER BY 
    1 DESC, 2 ASC
```
