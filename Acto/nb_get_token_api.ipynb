```python
import os
import time
import requests

# ── Configuração dos tenants ──────────────────────────────────
TOKEN_URL = "https://app-shared-prd-apiloginunico-002.codeciphers.com/ccloginunico/v2/Token"

TENANTS = {
    "santos": {
        "base_url":    "https://gestaosantosdigital.acto.net.br",
        "app_id":      "86BF9FC6-78AD-4A65-89E8-8C91F8EAC43D",
        "param_login": "5103",
        "param_user":  "CodCliente",
        "env_user":    "ACTO_SANTOS_USER",
        "env_senha":   "ACTO_SANTOS_SENHA",
    },

    "santos_obras": {
        "base_url":    "https://gestaoaprovasantos.acto.net.br",
        "app_id":      "86BF9FC6-78AD-4A65-89E8-8C91F8EAC43D",
        "param_login": "3997",
        "param_user":  "CodCliente",
        "env_user":    "ACTO_OSASCO_USER",
        "env_senha":   "ACTO_OSASCO_SENHA",
    },

    "osasco": {
        "base_url":    "https://gestaoosascodigital.acto.net.br",
        "app_id":      "86BF9FC6-78AD-4A65-89E8-8C91F8EAC43D",
        "param_login": "4861",
        "param_user":  "CodCliente",
        "env_user":    "ACTO_OSASCO_USER",
        "env_senha":   "ACTO_OSASCO_SENHA",
    },

    "maua": {
        "base_url":    "https://gestaomaua.acto.net.br",
        "app_id":      "86BF9FC6-78AD-4A65-89E8-8C91F8EAC43D",
        "param_login": "3736",
        "param_user":  "CodCliente",
        "env_user":    "ACTO_OSASCO_USER",
        "env_senha":   "ACTO_OSASCO_SENHA",
    },
}

# ── Cache em memória ──────────────────────────────────────────
_token_cache: dict = {}

def get_acto_token(tenant: str, env_user: str, env_senha: str) -> str:
    cached = _token_cache.get(tenant)
    if cached and cached["expires_at"] > time.time() + 60:
        return cached["token"]

    cfg = TENANTS[tenant]
    user  = env_user
    senha = env_senha

    if not user or not senha:
        raise EnvironmentError(f"Credenciais não encontradas para tenant '{tenant}'. "
                               f"Verifique o acto.env ou os.environ.")

    resp = requests.post(
        TOKEN_URL,
        headers={
            "accept":        "application/json",
            "content-type":  "application/x-www-form-urlencoded",
            "app_id":        cfg["app_id"],
            "param_login":   cfg["param_login"],
            "param_user":    cfg["param_user"],
            "authorization": "Bearer null",
        },
        data={
            "username":   user,
            "password":   senha,
            "grant_type": "password",
            "tipoLogin":  "0",
        },
        timeout=30,
    )
    resp.raise_for_status()
    data  = resp.json()
    token = data.get("access_token") or data.get("token")

    if not token:
        raise ValueError(f"Token não encontrado. Chaves recebidas: {list(data.keys())}")

    _token_cache[tenant] = {
        "token":      token,
        "expires_at": time.time() + data.get("expires_in", 3600),
    }
    return token
```
