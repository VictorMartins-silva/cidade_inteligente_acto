```python
import pandas as pd
import numpy as np


def consolidar_registros(df_reg01, df_reg02, df_reg03, df_reg04, df_reg07):
    # Selecionar colunas
    cols_reg1 = [
        "cod_est_cadastral_fam",
        "dat_atualizacao_familia",
        "dat_cadastramento_fam",
        "nom_logradouro_fam",
        "num_logradouro_fam",
        "nom_localidade_fam",
        "num_cep_logradouro_fam",
        "des_complemento_fam",
        "des_complemento_adic_fam",
        "nom_tip_logradouro_fam",
        "cod_familiar_fam",
        "vlr_renda_media_fam",
        "txt_referencia_local_fam",
        "cod_forma_coleta_fam",
    ]
    cols_reg2 = [
        "cod_familiar_fam", "qtd_comodos_domic_fam", "cod_agua_canalizada_fam", "cod_escoa_sanitario_domic_fam"
    ]
    cols_reg3 = [
        "cod_familiar_fam",
        "nom_centro_assist_fam",
        "qtd_familias_domic_fam",
        "qtd_pessoas_domic_fam",
        "val_desp_energia_fam",
        "val_desp_agua_esgoto_fam",
        "val_desp_alimentacao_fam",
    ]
    cols_reg4 = [
        "cod_familiar_fam",
        "dta_cadastramento_memb",
        "num_membro_fmla",
        "nom_pessoa",
        "dta_nasc_pessoa",
        "cod_sexo_pessoa",
        "filiacao_1_nom_completo_mae_pessoa",
        "num_nis_pessoa_atual",
        "cod_parentesco_rf_pessoa",
        "cod_est_cadastral_memb",
        "dta_atual_memb",
        "chv_nat_pes_atual",
    ]
    cols_reg7 = [
        "cod_familiar_fam",
        "num_membro_fmla",
        "ind_frequenta_escola_memb",
        "cod_curso_frequenta_memb",
        "cod_concluiu_frequentou_memb",
        "cod_curso_frequentou_pessoa_memb",
        "nom_escola_memb",
        "cod_sabe_ler_escrever_memb",
    ]

    df_reg01 = df_reg01.loc[:, cols_reg1].copy()
    df_reg02 = df_reg02.loc[:, cols_reg2].copy()
    df_reg03 = df_reg03.loc[:, cols_reg3].copy()
    df_reg04 = df_reg04.loc[:, cols_reg4].copy()
    df_reg07 = df_reg07.loc[:, cols_reg7].copy()

    # Consolidar tabelas, joins
    df = (
        df_reg01.merge(df_reg02, on="cod_familiar_fam", how="inner")
        .merge(df_reg03, on="cod_familiar_fam", how="inner")
        .merge(df_reg04, on="cod_familiar_fam", how="inner")
        .merge(df_reg07, on=["cod_familiar_fam", "num_membro_fmla"], how="inner")
    )

    return df


def aplicar_dicionarios(df):
    # aplicar dicionários
    cod_sexo_pessoa = {"1": "Masculino", "2": "Feminino"}
    df["desc_cod_sexo_pessoa"] = df["cod_sexo_pessoa"].map(cod_sexo_pessoa)

    cod_forma_coleta_fam = {"1": "Sem visita domiciliar", "2": "Com visita domiciliar"}
    df["desc_cod_forma_coleta_fam"] = df["cod_forma_coleta_fam"].map(
        cod_forma_coleta_fam
    )

    cod_parentesco_rf_pessoa = {
        "01": "Pessoa responsável pela Unidade Familiar - RF",
        "02": "Cônjuge ou companheiro(a)",
        "03": "Filho(a)",
        "04": "Enteado(a)",
        "05": "Neto(a) ou bisneto(a)",
        "06": "Pai ou mãe",
        "07": "Sogro(a)",
        "08": "Irmão ou irmã",
        "09": "Genro ou nora",
        "10": "Outro parente",
        "11": "Não parente",
    }
    df["desc_cod_parentesco_rf_pessoa"] = df["cod_parentesco_rf_pessoa"].map(
        cod_parentesco_rf_pessoa
    )

    cod_agua_canalizada_fam = {
        "1": "Sim",
        "2": "Não",
    }
    df["desc_cod_agua_canalizada_fam"] = df["cod_agua_canalizada_fam"].map(
        cod_agua_canalizada_fam
    )

    cod_escoa_sanitario_domic_fam = {
        "1": "Rede coletora de esgoto ou pluvial",
        "2": "Fossa séptica",
        "3": "Fossa rudimentar",
        "4": "Vala a céu aberto",
        "5": "Direto para um rio, lago ou mar",
        "6": "Outra forma",
    }
    df["desc_cod_escoa_sanitario_domic_fam"] = df[
        "cod_escoa_sanitario_domic_fam"
    ].map(cod_escoa_sanitario_domic_fam)

    cod_sabe_ler_escrever_memb = {"1": "Sim", "2": "Não"}
    df["desc_cod_sabe_ler_escrever_memb"] = df[
        "cod_sabe_ler_escrever_memb"
    ].map(cod_sabe_ler_escrever_memb)

    ind_frequenta_escola_memb = {
        "1": "Sim, rede pública",
        "2": "Sim, rede particular",
        "3": "Não, já frequentou",
        "4": "Nunca frequentou",
    }
    df["desc_ind_frequenta_escola_memb"] = df[
        "ind_frequenta_escola_memb"
    ].map(ind_frequenta_escola_memb)

    cod_curso_frequenta_memb = {
        "01": "Creche",
        "02": "Pré-escola (exceto CA)",
        "03": "Classe de Alfabetização - CA",
        "04": "Ensino Fundamental regular (duração 8 anos)",
        "05": "Ensino Fundamental regular (duração 9 anos)",
        "06": "Ensino Fundamental especial",
        "07": "Ensino Médio regular",
        "08": "Ensino Médio especial",
        "09": "Ensino Fundamental EJA - séries iniciais (Supletivo - 1ª a 4ª)",
        "10": "Ensino Fundamental EJA - séries finais (Supletivo - 5ª a 8ª)",
        "11": "Ensino Médio EJA (Supletivo)",
        "12": "Alfabetização para adultos",
        "13": "Superior, Aperfeiçoamento, Especialização, Mestrado, Doutorado",
        "14": "Pré-vestibular",
    }
    df["desc_cod_curso_frequenta_memb"] = df["cod_curso_frequenta_memb"].map(
        cod_curso_frequenta_memb
    )

    cod_curso_frequentou_pessoa_memb = {
        "01": "Creche",
        "02": "Pré-escola (exceto CA)",
        "03": "Classe de Alfabetização - CA",
        "04": "Ensino Fundamental 1ª a 4ª séries, Elementar (Primário), Primeira fase do 1º grau",
        "05": "Ensino Fundamental 5ª a 8ª séries, Médio 1º ciclo (Ginasial), Segunda fase do 1º grau",
        "06": "Ensino Fundamental (duração 9 anos)",
        "07": "Ensino Fundamental Especial",
        "08": "Ensino Médio, 2º grau, Médio 2º ciclo (Científico, Clássico, Técnico, Normal)",
        "09": "Ensino Médio Especial",
        "10": "Ensino Fundamental EJA - séries iniciais (Supletivo - 1ª a 4ª)",
        "11": "Ensino Fundamental EJA - séries finais (Supletivo - 5ª a 8ª)",
        "12": "Ensino Médio EJA (Supletivo)",
        "13": "Superior, Aperfeiçoamento, Especialização, Mestrado, Doutorado",
        "14": "Alfabetização para adultos (Mobral, etc.)",
        "15": "Nenhum",
    }
    df["desc_cod_curso_frequentou_pessoa_memb"] = df[
        "cod_curso_frequentou_pessoa_memb"
    ].map(cod_curso_frequentou_pessoa_memb)
    return df


def tratar_datas(df):
    colunas_data = ["dat_cadastramento_fam", "dat_atualizacao_familia"]
    for col in colunas_data:
        df[col] = pd.to_datetime(df[col], format="%d%m%Y")
    return df


def tratar_exportacao_pruc(df):
    # alterar tipo renda media
    df["int_vlr_renda_media_fam"] = df["vlr_renda_media_fam"].astype(int)

    # calcular tamanho das familias
    df["n_pessoas_fam"] = df.groupby("cod_familiar_fam")[
        "cod_familiar_fam"
    ].transform("count")

    # calcular renda per capita
    df["vlr_renda_media_fam"] = (
        df["int_vlr_renda_media_fam"] / 100
    )

    # # EXPORTAR BASE FINAL PARA CONSUMO NO DASH
    colunas_final = [
        "num_cep_logradouro_fam",
        "dat_cadastramento_fam",
        "cod_familiar_fam",
        "vlr_renda_media_fam",
        "desc_cod_sexo_pessoa",
        "desc_cod_forma_coleta_fam",
        "desc_cod_parentesco_rf_pessoa",
        "qtd_pessoas_domic_fam",
        "qtd_comodos_domic_fam",
        "qtd_familias_domic_fam",
        "val_desp_energia_fam",
        "val_desp_agua_esgoto_fam",
        "val_desp_alimentacao_fam",
        "desc_cod_agua_canalizada_fam",
        "desc_cod_escoa_sanitario_domic_fam",
        "desc_cod_sabe_ler_escrever_memb",
        "desc_ind_frequenta_escola_memb",
        "desc_cod_curso_frequentou_pessoa_memb",
        "desc_cod_curso_frequenta_memb",
        "n_pessoas_fam",
    ]

    df_ativos = df.loc[
        (df["cod_est_cadastral_fam"] != "4")
        & (df["cod_est_cadastral_memb"] != "4"),
        colunas_final
    ].reset_index(drop=True).copy()

    return df_ativos
```

```python
def main():

    print("Lendo blocos do CadÚnico...")
    df_reg01 = spark.table("bronze_cad_unico_reg01").toPandas()
    df_reg01['versao_arquivo'] = pd.to_datetime(df_reg01['versao_arquivo'])
    versao = df_reg01['versao_arquivo'].max()
    df_reg01 = df_reg01.loc[df_reg01['versao_arquivo'] == versao].copy()

    df_reg02 = spark.table("bronze_cad_unico_reg02").toPandas()
    df_reg02['versao_arquivo'] = pd.to_datetime(df_reg02['versao_arquivo'])
    df_reg02 = df_reg02.loc[df_reg02['versao_arquivo'] == versao].copy()

    df_reg03 = spark.table("bronze_cad_unico_reg03").toPandas()
    df_reg03['versao_arquivo'] = pd.to_datetime(df_reg03['versao_arquivo'])
    df_reg03 = df_reg03.loc[df_reg03['versao_arquivo'] == versao].copy()

    df_reg04 = spark.table("bronze_cad_unico_reg04").toPandas()
    df_reg04['versao_arquivo'] = pd.to_datetime(df_reg04['versao_arquivo'])
    df_reg04 = df_reg04.loc[df_reg04['versao_arquivo'] == versao].copy()

    df_reg07 = spark.table("bronze_cad_unico_reg07").toPandas()
    df_reg07['versao_arquivo'] = pd.to_datetime(df_reg07['versao_arquivo'])
    df_reg07 = df_reg07.loc[df_reg07['versao_arquivo'] == versao].copy()

    print("Consolidando os registros do CadÚnico...")
    df = consolidar_registros(
        df_reg01, df_reg02, df_reg03, df_reg04, df_reg07
    )

    print("Aplicando dicionários de códigos do CadÚnico...")
    df = aplicar_dicionarios(df)

    print("Aplicando tratamentos para exportação do CadÚnico...")
    df = tratar_datas(df)

    dffinal = tratar_exportacao_pruc(df)

    return dffinal

dffinal = main()
```

```python
bairros = pd.read_csv("/lakehouse/default/Files/cadastro_unico/cep_bairros.csv", sep=";", dtype={"cep_formatado": str})

def formatar_cep(cep):
    # Garante que o CEP seja uma string e tenha 8 dígitos
    cep = str(cep).zfill(8)
    return f"{cep[:5]}-{cep[5:]}"

dffinal["cep_formatado"] = dffinal["num_cep_logradouro_fam"].apply(formatar_cep)
df = pd.merge(dffinal, bairros, on="cep_formatado", how="left")
```

```python
cad_unico = df.copy()

# cod_familiar_fam
cod_familiar_fam = cad_unico[["cod_familiar_fam"]].drop_duplicates()

# cod_familiar_fam_2025
cad_unico["ano_cadastramento"] = pd.to_datetime(
    cad_unico["dat_cadastramento_fam"]
).dt.year
cod_familiar_fam_2025 = cad_unico.query("ano_cadastramento == 2026")[
    ["cod_familiar_fam"]
].drop_duplicates()

# renda_per_capita_fam
cad_unico['vlr_renda_per_capita_fam'] = cad_unico['vlr_renda_media_fam']
renda_per_capita_fam = cad_unico[["cod_familiar_fam", "vlr_renda_per_capita_fam"]].drop_duplicates()

# n_pessoas_fam
n_pessoas_fam = (
    cad_unico[["cod_familiar_fam", "n_pessoas_fam", "qtd_pessoas_domic_fam"]]
    .drop_duplicates()
)

# agua_canalizada_fam
agua_canalizada_fam = (
    cad_unico[["desc_cod_agua_canalizada_fam", "cod_familiar_fam"]]
    .drop_duplicates()
)

# qtd_comodos_domic_fam
qtd_comodos_domic_fam = (
    cad_unico[["cod_familiar_fam", "qtd_comodos_domic_fam"]]
    .drop_duplicates()
)

# sabe_ler_escrever_memb
sabe_ler_escrever_memb = cad_unico["desc_cod_sabe_ler_escrever_memb"].value_counts(0).reset_index()

# sexo_pessoa
sexo_pessoa = cad_unico["desc_cod_sexo_pessoa"].value_counts(0).reset_index()

# forma_coleta
cad_unico["desc_cod_forma_coleta_fam"] = cad_unico["desc_cod_forma_coleta_fam"].fillna(
    "Não informado"
)
cad_unico["dat_cadastramento_fam"] = pd.to_datetime(cad_unico["dat_cadastramento_fam"])
cad_unico["ano_cadastramento"] = cad_unico["dat_cadastramento_fam"].dt.year
forma_coleta = (
    cad_unico.query("ano_cadastramento >= 2000")[
        ["ano_cadastramento", "cod_familiar_fam", "desc_cod_forma_coleta_fam"]
    ]
    .drop_duplicates()
    .drop(columns="cod_familiar_fam")
    .groupby(["ano_cadastramento", "desc_cod_forma_coleta_fam"])
    .size()
    .reset_index(name="n_familias")
)

# parentesco
parentesco = (
    cad_unico["desc_cod_parentesco_rf_pessoa"]
    .value_counts(ascending=False)
    .reset_index()
    .sort_values("count")
)

# indicadores_bairros
colunas_numericas = [
    "vlr_renda_media_fam",
    "qtd_pessoas_domic_fam",
    "qtd_comodos_domic_fam",
    "qtd_familias_domic_fam",
    "val_desp_energia_fam",
    "val_desp_agua_esgoto_fam",
    "val_desp_alimentacao_fam",
]

for col in colunas_numericas:
    cad_unico[col] = (
        cad_unico[col]
        .astype(str)
        .str.strip()
        .str.replace(",", ".", regex=False)
        .replace({"": None, "nan": None, " ": None})
    )
    cad_unico[col] = pd.to_numeric(cad_unico[col], errors="coerce")
cad_unico["bairro"] = cad_unico["bairro"].str.replace("km 18", "Km 18")
indicadores_bairros = (
    cad_unico.groupby("bairro", as_index=False)
    .agg(
        {
            "cod_familiar_fam": "nunique",
            "vlr_renda_per_capita_fam": "mean",
            "qtd_pessoas_domic_fam": "mean",
            "qtd_comodos_domic_fam": "mean",
            "qtd_familias_domic_fam": "mean",
            "val_desp_energia_fam": "mean",
            "val_desp_agua_esgoto_fam": "mean",
            "val_desp_alimentacao_fam": "mean",
        }
    )
    .round(2)
    .sort_values("cod_familiar_fam", ascending=False)
)

# escoa_sanitario_fam
escoa_sanitario_fam = (
    cad_unico[["desc_cod_escoa_sanitario_domic_fam", "cod_familiar_fam"]]
    .drop_duplicates()
)
```

```python
tabelas_painel = {
    "cod_familiar_fam": cod_familiar_fam,
    "cod_familiar_fam_2025": cod_familiar_fam_2025,
    "renda_per_capita_fam": renda_per_capita_fam,
    "n_pessoas_fam": n_pessoas_fam,
    "agua_canalizada_fam": agua_canalizada_fam,
    "qtd_comodos_domic_fam": qtd_comodos_domic_fam,
    "sabe_ler_escrever_memb": sabe_ler_escrever_memb,
    "sexo_pessoa": sexo_pessoa,
    "forma_coleta": forma_coleta,
    "parentesco": parentesco,
    "indicadores_bairros": indicadores_bairros,
    "escoa_sanitario_fam": escoa_sanitario_fam,
}

for nome, tabela in tabelas_painel.items():
    tabela_spark = spark.createDataFrame(tabela)
    (
        tabela_spark
        .write
        .mode("overwrite")
        .format("delta")
        .option("overwriteSchema", "true")
        .saveAsTable(f"gold_cad_unico_{nome}")
    ) 
```
