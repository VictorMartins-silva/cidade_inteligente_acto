```python
TOKEN = TOKEN_SANTOS

# ── Caminhos ──────────────────────────────────────────────────
BASE_PATH       = "/lakehouse/default/Files"
PAYLOAD_PATH    = BASE_PATH + "/acto_gestao_api_payload/payload_santos_curso_motorista.json"
SILVER_PATH     = BASE_PATH + "/silver/santos_curso_motorista/"

# ── Extração via API ──────────────────────────────────────────
df_solicitacoes, df_etapas = extrair_tabela_acto_gestao(PAYLOAD_PATH, TOKEN)

print(f"Solicitações: {df_solicitacoes.shape}")
print(f"Etapas      : {df_etapas.shape}")
```

```python
# ── Validação rápida antes de salvar ──────────────────────────
print("=== Solicitações (primeiras linhas) ===")
display(df_solicitacoes.head(5))

print("\n=== Etapas (primeiras linhas) ===")
display(df_etapas.head(5))
```

```python
# ── Salva na Silver (raw, sem transformação) ──────────────────
import os
os.makedirs(SILVER_PATH, exist_ok=True)

df_solicitacoes.to_parquet(SILVER_PATH + "silver_solicitacoes.parquet", index=False)
df_etapas.to_parquet(      SILVER_PATH + "silver_etapas.parquet",       index=False)

print(f"✓ silver_solicitacoes.parquet salvo — {df_solicitacoes.shape[0]} linhas")
print(f"✓ silver_etapas.parquet       salvo — {df_etapas.shape[0]} linhas")
print(f"  Pasta: {SILVER_PATH}")
```
